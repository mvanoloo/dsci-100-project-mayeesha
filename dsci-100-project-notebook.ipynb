```python
import altair as alt
bball_plot1=alt.Chart(bball_data,title='Womens NCAA Basketball Tourney Wins').mark_bar().encode(
    x=alt.X('tourney_wins').title('Number of Tourney Wins'),
    y=alt.Y('count()').title("Count of Number of Tourney Wins")
)
bball_plot1
```

```python
bball_plot2=alt.Chart(bball_data,title="NCAA Women's Basketball Conferences").mark_bar().encode(
    x=alt.X('conference').title('Conference'),
    y=alt.Y('count()').title('Count of Conference')
)
bball_plot2
```
